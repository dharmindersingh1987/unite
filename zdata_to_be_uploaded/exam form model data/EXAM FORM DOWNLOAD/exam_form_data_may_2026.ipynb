```python
from pathlib import Path
import pandas as pd

folder = Path("./exam_form_data_may_2026")

df = pd.concat(
    [
        pd.read_excel(file).assign(source_file=file.stem)
        for file in folder.glob("*.xlsx")
    ],
    ignore_index=True
)
```

```python
df.to_csv("/home/dharm/exam_form_data_may_2026_compiled_full.csv", index=False)
```

```python
df["schemeTitle"] = (
    df["schemeTitle"]
    .str.split(",", n=2)
    .str[:2]
    .str.join(",")
)
```
