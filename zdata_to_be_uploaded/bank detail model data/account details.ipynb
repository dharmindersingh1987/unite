```python
import pandas as pd
```

```python
pbps = pd.read_html(
    "ee.html",
    index_col=0,
    converters={"acc": str}
)[0]
```

```python
def strip_strings(pbps):
    return pbps.apply(
        lambda col: col.map(
            lambda x: x.strip() if isinstance(x, str) else x
        )
    )
pbps=pbps.pipe(strip_strings).copy()
```
